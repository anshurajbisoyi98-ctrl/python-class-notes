```python
class Solution:
    def isPalindrome(self, s: str) -> bool:
      s=s.lower()
      new_st=""
      for i in s:
        if(i>='a'and i<='z')or (i>='0'and i<='9'):
          new_st+=i
      return new_st==new_st[::-1]
```
