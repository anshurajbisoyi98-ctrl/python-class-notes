```python
matstock1=np.random.randint(1,90,size=(5,5))
matdemand2=np.random.randint(1,90,size=(5,5))
print(matstock1)
print(matdemand2)
remainingstock = matstock1-matdemand2
print(remainingstock)
print(" ")
remainingstockupdated=np.where(remainingstock<0,0,remainingstock)
print(remainingstockupdated)
print(" ")
totalshortage=np.sum(np.where(remainingstock<0,remainingstock,0))
print(-totalshortage)
print(" ")
productswithzerostock=np.sum(np.where(remainingstockupdated==0,1,0),axis=0)
print(productswithzerostock)
print(" ")
sorteachremainingstock=np.sort(remainingstockupdated,1)
print(sorteachremainingstock)
print(" ")
```
