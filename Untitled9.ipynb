```python
import pandas as pd
df1= pd.DataFrame({
    'id': [1,2,3,4],
    'Name': ['raja','rani','rahul','rhea'],
    'Age': [21,22,23,24]
})

df2= pd.DataFrame({
    'id': [3,4,5,6],
    'Name': ['rahul','rhea','ravi','reena'],
    'Age': [21,22,23,24]
})

df3 = pd.merge(df1,df2,on='id',how='inner')
display(df3)
df3.rename(columns={'name_x':'table_name1'},inplace=True)#important functions remeber a;;; these and check for documentation to go in depth
display(df3)
```

```python
df3 = pd.merge(df1,df2,on='id',how='outer')
display(df3)
```

```python
df3 = pd.merge(df1,df2,on='id',how='left')
display(df3)
```

```python
display(df1)
display(df2)
```

```python
df3 = pd.merge(df1,df2,on='id',how='right')
display(df3)
```
